```python
import numpy as np
i=np.array([[1,2],[4,6]])
j=np.array([[6,5],[7,6]])

print("addition:",i+j)
print("\n")
print("substraction:",i-j)
print("\n")
print("multiplication:",i*j)
print("\n")
print("division:",i/j)
print("\n")
print("matrix multiplication",np.dot(i,j))
print("\n")
print("Transpose i:",i.T)
print("\n")
print("Transpose j:",j.T)


```
